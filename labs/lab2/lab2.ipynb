```python
import numpy as np
from boundaries import BoundaryType
from entities import SinusoidalSource, WaveSimConfig
from visual import plot_amplitude_map
from wave_equation import run_simulation
```

```python
configs = [
    WaveSimConfig(
        omega=5 * np.pi
    ),
    WaveSimConfig(
        omega=10 * np.pi
    ),
    WaveSimConfig(
        omega=20 * np.pi
    )
]

for config in configs:
    source = SinusoidalSource(
        amplitude=config.amplitude,
        omega=config.omega
    )
    A_max = run_simulation(source, config)
    plot_amplitude_map(
        A_max=A_max,
        cmap=config.cmap,
        amplitude=config.amplitude,
        omega=config.omega
    )
```

```python
configs = [
    WaveSimConfig(
        omega=5 * np.pi,
        source_positions=[(100, 10), (100, 190)]
    ),
    WaveSimConfig(
        omega=10 * np.pi,
        source_positions=[(100, 10), (100, 190)]
    ),
    WaveSimConfig(
        omega=20 * np.pi,
        source_positions=[(100, 10), (100, 190)]
    )
]

for config in configs:
    omega = 7 * np.pi
    sources = [
        SinusoidalSource(
            amplitude=config.amplitude,
            omega=config.omega
        ),
        SinusoidalSource(
            amplitude=config.amplitude,
            omega=omega
        )
    ]
    A_max = run_simulation(sources, config)
    plot_amplitude_map(
        A_max=A_max,
        cmap=config.cmap,
        amplitude=config.amplitude,
        omega=f"{config.omega} and {omega}"
    )
```
